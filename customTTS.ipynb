```python
# Cell 1 - imports
from reachy_sdk import ReachySDK
from gtts import gTTS
import pygame
import io
```

```python
# Cell 2 - connect to Reachy
reachyip = '152.23.165.203'  # Replace with your Reachy's IP address
reachy = ReachySDK(host=reachyip)  
```

```python
# Cell 3 - TTS function
def speak(text, lang='en'):
    tts = gTTS(text=text, lang=lang)
    buf = io.BytesIO()
    tts.write_to_fp(buf)
    buf.seek(0)
    pygame.mixer.init()
    pygame.mixer.music.load(buf, 'mp3')
    pygame.mixer.music.play()
    while pygame.mixer.music.get_busy():
        pygame.time.Clock().tick(10)
```

```python
# Cell 4 - test
speak("Hello, how are you today")
```
